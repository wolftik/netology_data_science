```python
# Концептуальный пример
from llama_cpp import Llama

# Указываешь путь к модели из LM Studio
model_path = r"E:\LLM-models\lmstudio_models\lmstudio-community\gemma-4-E4B-it-GGUF\gemma-4-E4B-it-Q4_K_M.gguf"

# Создаешь модель
llm = Llama(
    model_path=model_path,
    n_gpu_layers=-1,  # 0 = полностью на CPU, -1 = на GPU если есть
    n_ctx=12048,      # контекстное окно
    verbose=True 
)

# Используешь
output = llm(
    "Привет, что умеешь?",  
    max_tokens=100,         # макс. токенов в ответе
    temperature=0.7         # креативность (0-1)
)

print(output['choices'][0]['text'])
```
