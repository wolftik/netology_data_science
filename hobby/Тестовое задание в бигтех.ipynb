## Задание 2 (python)
Дана строка из заглавных букв A-Z: AABCCCDDEFFFF

Нужно посчитать сколько раз каждая буква встречается подряд:
Если символ встречается один раз, он остается неизменным;
Если символ встречается более одного раза, к нему добавляется число повторений.
Результат: A2BC3D2EF4

```python
s = 'AABCCCDDEFFFF'
new_s=''    

for i in s:
    t = s.count(i)
    if t > 1:
        new_s += i+str(t)
        
print(new_s)
```
