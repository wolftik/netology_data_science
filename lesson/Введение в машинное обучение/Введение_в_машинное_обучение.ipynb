<h2><b>Что такое машинное обучение</b></h2>

__Обучение__ — приобретение необходимой функциональности посредством опыта.

__Машинное обучение__ — процесс, в результате которого машина способна показывать поведение, которое в неё не было явно запрограммировано. Например, функционал калькулятора полностью запрограммирован, а в камерах слежения за ПДД запрограммирован алогритм детекции машин.

Компьютерная программа обучается из опыта __E__ в классе задач __T__ с мерой качества __P__, если качество измеренное с помощью __P__ в классе задач __T__ увеличивается по мере увеличения опыта __E__». Том Митчел

- Задача: игра в шашки / шахматы / го
- Мера: процент побед
- Опыт: игра программы против себя


- Задача: рекомендация товаров/услуг/видео
- Мера: процент успешных рекомендаций
- Опыт: список товаров, просмотренных/купленных/оценённых пользователями

__Машинное обучение__ – это мультидисциплинарная область, возникшая и развивающаяся на базе таких наук как прикладная статистика, искусственный интеллект, теория баз данных и др.
* Единого определения машинного обучения на сегодняшний день нет. Его можно охарактеризовать как формирование алгоритмов, позволяющих находить в «сырых» данных ранее неизвестных, неочевидных, объективных, практически полезных и доступных интерпретаций знаний, необходимых для принятия решений в различных сферах человеческой деятельности. __Неочевидных__ – это значит, что найденные закономерности не обнаруживаются стандартными методами обработки информации или экспертным путём. __Объективных__ – это значит, что найденные закономерности будут полностью соответствовать действительности, в отличие от экспертного мнения, которое всегда является субъективным.
* Средства машинного обучения, в отличие от статистических, могут опираться на слабо структурированные и (или) неполные данные.
* Очевидно, что машинное обучение не стоит применять тогда, когда зависимость между переменными уже известна. Например, зависимость силы от массы и ускорения.

<h2><b>Основные подходы для обучения алгоритмов</b></h2>
В современном машинном обучении можно выделить два больших направления.

* __Классическое машинное обучение__:
    * Включает в себя большинство популярных алгоритмов
    * Может иметь дело с разнородными данными (числа, строки, булевы значения, категориальные данные и т.д.)
    * Произвольные размеры данных – от очень малых (десятки записей) до миллионов записей и более
    * Сильный упор на математику, статистику, теорию вероятностей
* __Глубокое обучение (deep learning)__:
    * Основывается практически исключительно на нейронных сетях специальной структуры
    * Решает задачи обработки очень больших объёмов однородных данных (изображения, видео, звук, текст), также включает в себя задачу генерирования новых данных
    * Огромные модели вычислений (миллионы параметров и гигабайты выборок – обычное дело) и упор на высокопроизводительные вычисления (GPU, кластеры, суперкомпьютеры)
    * Больше инженерии, чем строгой теории
    * Эта область обладает определённой спецификой

Задачи машинного обучения делятся на две основные группы: __обучение с учителем (supervised learning)__ и __обучение без учителя (unsupervised learning)__.
Есть и другие типы задач (частичное обучение, обучение с подкреплением), но они специфичны.

__Обучение с учителем__

Имеется множество объектов __X__ и множество возможных ответов __Y__, заданные на конечной обучающей выборке. На основе этих данных требуется восстановить целевую зависимость, то есть построить алгоритм, способный для любого объекта, подобного имеющимся в выборке, выдать достаточно точный ответ. Другими словами, нужно построить отображение __X → Y__, с приемлемым качеством позволяющее определить значение целевого атрибута __Y__ для любого __X__, взятого из того же распределения, что и объекты обучающей выборки, в том числе для тех __X__, которые в точности не представлены в выборке.

__Обучение без учителя__

Обучение без учителя используется, когда имеется множество объектов __X__, а целевого атрибута не задано. Цель обучения без учителя – позволить алгоритму самостоятельно разобраться со структурой распределения объектов __X__, например, сгруппировать их в кластеры по
отношению сходства и/или выделить из массы объектов __X__ аномалии (объекты, сильно отличающиеся от других).

<h2><b>Sklearn. Линейная регрессия</b></h2>

Рассмотрим одну из самых простых и в то же время распространённых моделей – линейную регресиию. Реализовывать её с нуля нет необходимости, так как за нас это уже сделали. Достаточно импортировать модель из нужной библиотеки, научиться использовать и улучшать её.

Для обучения моделей нужны какие-то данные. Давайте сгенерируем случайные с помощью библиотеки __numpy__.

```python
import numpy as np
X = np.random.randint(100, size=(7, 3))
X
```

Построим __y__ по такой формуле:

$$y = 25t + 100s - 42z + 81 + \varepsilon$$

Где $\varepsilon$ – это какой-то шум с нормальным распределением с дисперсией 10, чтобы линейная формула не была уж совсем точной.

```python
y = np.dot(X, np.array([25, 100, -42])) + 81 + np.random.normal(0, 10, len(X))
y
```

Приступим к обучению.

```python
from sklearn.linear_model import LinearRegression # импорт модели
model = LinearRegression() # инициализация модели
```

```python
model.fit(X, y) # обучение модели
```

```python
coefs = model.coef_ # коэффициенты перед t, s, z, которые подобрала модель
coefs
```

```python
const = model.intercept_ # свободный член уравнения
const
```

По сути мы хотим подобрать числа $a_0, a_1, a_2, a_3$ для вот такой модели:

$$f(x_1, x_2, x_3) = a_0 + a_1 x_1 + a_2 x_2 + a_3 x_3$$

Мы хотим подобрать их так, чтобы функция потерь на наших данных была минимальна. В LinearRegression используют функцию потерь MSE - сумму квадратов отклонений от настоящего значения.

То есть задача такая:

$$\sum\limits_{i=1}^{n}(f(x_{i1}, x_{i2}, x_{i3}) - y_i)^2 \rightarrow \min$$


$$\sum\limits_{i=1}^{n}(a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i)^2 \rightarrow \min$$

Где n - это количество входных данных. Давайте рассмотрим эту сумму как функцию от 4 переменных $a_0, a_1, a_2, a_3$, которую нам нужно минимизировать. А числа $x_{ij}$ и $y_i$, получается, будут обычными константами.

$$MSE(a_0, a_1, a_2, a_3) \rightarrow \min$$

Давайте посчитаем частную производную по каждой координате.

Начнём с координаты $a_1$.

$$MSE'_{a_1}(a_0, a_1, a_2, a_3) = \sum\limits_{i=1}^{n}((a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i)^2)'_{a_1}=$$

Раскрываем квадрат, вынося отдельно члены, которые делятся на $a_1^2$, $a_1$ и $1$.

$$= \sum\limits_{i=1}^{n}(x_{i1}^2a_1^2 + 2x_{i1}(a_0 + a_2 x_{i2} + a_3 x_{i3} - y_i)a_1 + (a_0 + a_2 x_{i2} + a_3 x_{i3} - y_i)^2)'_{a_1}=$$

Считаем производную, одна из скобок при этом обнуляется:

$$= \sum\limits_{i=1}^{n}(2x_{i1}^2a_1 + 2x_{i1}(a_0 + a_2 x_{i2} + a_3 x_{i3} - y_i))=$$

Теперь вынесем $2$ и $x_{i1}$

$$= 2\sum\limits_{i=1}^{n}(x_{i1}a_1 + a_0 + a_2 x_{i2} + a_3 x_{i3} - y_i)x_{i1}=$$

Заметим, что в скобках получилось очень простое выражение!

$$= 2\sum\limits_{i=1}^{n}(f(x_{i1}, x_{i2}, x_{i3}) - y_i)x_{i1}$$

Давайте приравняем все 4 производные (по $a_0, a_1, a_2, a_3$) нулю, тогда:

$$\sum\limits_{i=1}^{n}(a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i) = 0$$
$$\sum\limits_{i=1}^{n}(a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i)x_{i1} = 0$$
$$\sum\limits_{i=1}^{n}(a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i)x_{i2} = 0$$
$$\sum\limits_{i=1}^{n}(a_0 + a_1 x_{i1} + a_2 x_{i2} + a_3 x_{i3} - y_i)x_{i3} = 0$$

Давайте сгруппируем все выражения по $a_0, a_1, a_2, a_3$:

$$na_0 + (\sum\limits_{i=1}^{n}x_{i1})a_1 + (\sum\limits_{i=1}^{n}x_{i2})a_2 + (\sum\limits_{i=1}^{n}x_{i3})a_3= \sum\limits_{i=1}^{n}y_i$$

$$(\sum\limits_{i=1}^{n}x_{i1})a_0 + (\sum\limits_{i=1}^{n}x_{i1}^2)a_1 + (\sum\limits_{i=1}^{n}x_{i1}x_{i2})a_2 + (\sum\limits_{i=1}^{n}x_{i1}x_{i3})a_3= \sum\limits_{i=1}^{n}x_{i1}y_i$$

$$(\sum\limits_{i=1}^{n}x_{i2})a_0 + (\sum\limits_{i=1}^{n}x_{i1}x_{i2})a_1 + (\sum\limits_{i=1}^{n}x_{i2}^2)a_2 + (\sum\limits_{i=1}^{n}x_{i2}x_{i3})a_3= \sum\limits_{i=1}^{n}x_{i2}y_i$$

$$(\sum\limits_{i=1}^{n}x_{i3})a_0 + (\sum\limits_{i=1}^{n}x_{i1}x_{i3})a_1 + (\sum\limits_{i=1}^{n}x_{i2}x_{i3})a_2 + (\sum\limits_{i=1}^{n}x_{i3}^2)a_3= \sum\limits_{i=1}^{n}x_{i3}y_i$$

Мы получили красивую симметричную систему уравнений: 4 уравнения, 4 неизвестных. Если определитель матрицы коэффициентов не равен нулю, то у него есть ровно одно решение, и его мы умеем находить (методом Гаусса, например). Если определитель вдруг стал равен нулю, то решений либо 0, либо бесконечно.

У непрерывно-дифференцируемой функции, которая при стремлении по каждой координате к плюс или минус бесконечности сама стремится к плюс бесконечности, всегда существует глобальный минимум. В точке глобального минимума все производные как раз равны нулю. Следовательно, существует всегда хотя бы одно решение, и мы его найдём.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.simplefilter("ignore")
%matplotlib inline

sns.set(style="darkgrid")
```

Применяя метод линейной регрессии мы пытаемся подобрать коэффициенты для точек данных так, чтобы наша полученная функция была максимально похожа на искомую. По этапам мы оптимизируем коэффициенты, чтобы уменьшить функцию потерь. Но до достижения какого значения стоит продолжать оптимизацию? Стоит ли добиваться того, чтобы функция потерь была близка к 0? В таком случае мы можем столкнуться с эффектом переобучения (самая правая картинка).

```python
from sklearn.linear_model import LinearRegression

np.random.seed(36)
x = np.linspace(0, 1, 100)
y = np.cos(1.5 * np.pi * x)

x_objects = np.random.uniform(0, 1, size=30)
y_objects = np.cos(1.5 * np.pi * x_objects) + np.random.normal(scale=0.1, size=x_objects.shape)

from sklearn.preprocessing import PolynomialFeatures
fig, axs = plt.subplots(figsize=(16, 4), ncols=3)
for i, degree in enumerate([1, 4, 20]):
    X_objects = PolynomialFeatures(degree).fit_transform(x_objects[:, None])
    X = PolynomialFeatures(degree).fit_transform(x[:, None])
    regr = LinearRegression().fit(X_objects, y_objects)
    y_pred = regr.predict(X)
    axs[i].plot(x, y, label='Искомая функция')
    axs[i].scatter(x_objects, y_objects, label='Данные')
    axs[i].plot(x, y_pred, label='Предсказанная функция')
    if i == 0:
        axs[i].legend()
    axs[i].set_title("Степень полинома = %d" % degree)
    axs[i].set_xlabel("$x$")
    axs[i].set_ylabel("$f(x)$")
    axs[i].set_ylim(-2, 2)
```
