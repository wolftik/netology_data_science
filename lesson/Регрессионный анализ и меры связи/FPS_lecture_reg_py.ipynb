```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('FPS_lecture_df_0_height-weignt.xlsx')
df.head()
```

```python
# Формирование выборок
df_1 = df[df['Gender'] == 'Male']
df_2 = df[df['Gender'] == 'Female']

print('Выборкки имеют одинаковую длинну: ', len(df_1) == len(df_2))
```

```python
# Описательные статистики мужчин
df_1.describe()
```

```python
# Описательные статистики женщин
df_2.describe()
```

```python
# Визуализация пересечения интервалов веса мужчин и женщин
# более популярный и удобный вариант с distplot
sns.distplot(df_1['Weight'], 50)
sns.distplot(df_2['Weight'], 50)
plt.title('Плотность распределения Веса')
plt.legend(['Male', 'Female'])
plt.show()
```

```python
# актуальная к поддержке в библиотеке альтернатива через histplot
sns.histplot(df_1['Weight'], kde=True)
sns.histplot(df_2['Weight'], kde=True)
plt.title('Плотность распределения Веса')
plt.legend(['Male', 'Female'])
plt.show()
```

```python
# Функия бинарной классификации по признаку пола
def func_bin_gender(row):
 if row['Gender'] == 'Male':
    return 1
 elif row['Gender'] == 'Female':
    return 0
```

```python
# Добавление столбца бинарной классификации пола
df['Gender(bin)'] = df.apply(func_bin_gender, axis=1)
df
```

```python
# Корреляция Роста к Весу
df['Height'].corr(df['Weight'])
```

```python
# Корреляция Пола к Росту
df['Gender(bin)'].corr(df['Height'])
```

```python
# Корреляция Пола к Весу
df['Gender(bin)'].corr(df['Weight'])
```

```python
# Матрица корреляции средствами pandas
df[['ID', 'Height', 'Weight', 'Gender(bin)']].corr().style.background_gradient(cmap='coolwarm')
```

```python
# Матрица корреляции средствами визуализации seaborn (тепловая матрица)
sns.heatmap(df[['ID', 'Height', 'Weight', 'Gender(bin)']].corr(), annot = True, vmin=-1, vmax=1, center= 0, cmap= 'coolwarm')
```

## Построение Линейной регрессии

```python
# Формирование выборок Роста и Веса для Мужчин
X = df_1[['Height']]
y = df_1[['Weight']]
```

```python
from sklearn.linear_model import LinearRegression # Подключение модуля Линейной регрессии
regressor = LinearRegression()                    # Инициализация Линейной регрессии
regressor.fit(X, y)                               # Обучение Линейной регрессии
```

```python
# Коэффициент прямой
a = regressor.intercept_
a
```

```python
# Коэффициент наклона прямой
b = regressor.coef_
b
```

```python
Height_pred = 170                     # Значение Роста для получения Прогноза Веса
regressor.predict([[Height_pred]])    # Получение прогноза Линейной реграссии
```

```python
# Получение прогноза "в лоб" по уровнению прямой регрессии a + bx
a + b * Height_pred
```

Аналогичное построение регрессии для Женщин

```python
# Формирование выборок Роста и Веса для Женщин
X = df_2[['Height']]
y = df_2[['Weight']]

regressor = LinearRegression()                    # Инициализация Линейной регрессии
regressor.fit(X, y)                               # Обучение Линейной регрессии

# Коэффициент прямой
a = regressor.intercept_
# Коэффициент наклона прямой
b = regressor.coef_

Height_pred = 160                     # Значение Роста для получения Прогноза Веса
regressor.predict([[Height_pred]])    # Получение прогноза Линейной реграссии
```
