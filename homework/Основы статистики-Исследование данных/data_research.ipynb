```python
import pandas as pd
import statistics as st
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_excel('FPS_hw_1_df_1_bank_customer.xlsx')
df.head()
```

```python
df.describe()
```

```python
print('AGE statistic','\n')
print('Среднее:',st.mean(df['Age']),'\n')
print('Стандартное отклонение:',st.stdev(df['Age']),'\n')
print('Минимум:',df['Age'].min(),'\n')
print('Перцентили:','\n',df['Age'].quantile([0, 0.25, 0.50, 0.75, 1]),'\n')
print('Максимум:',df['Age'].max(),'\n')
```

```python
df['Age'].describe()
```

```python
# Гистограмма плотности распределения данных
plt.hist(df['Age'], 50, density=True)
plt.title('Распределение возраста клиентов банка')
plt.xlabel('Возраст')
plt.show()
```

## Вывод

Распределение возраста клиентов является мультимодальным, т.к. на гистограмме имеется несколько пиков. Также имеется выброс в районе 85 лет, есть клиент(ы) долгожитель(и).

```python
# График динамики значений
plt.plot(df['Age'])
plt.title('Динамика значений возраста клиентов банка')
plt.ylabel('Вес')
plt.show()
```
