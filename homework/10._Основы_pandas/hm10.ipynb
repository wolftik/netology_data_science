## Задание 1
Определите, какому фильму было выставлено больше всего оценок 5.0.

```python
movies_df = pd.read_csv('Файлы к заданию №1/movies.csv')
ratings_df = pd.read_csv('Файлы к заданию №1/ratings.csv')
movies_ra_df  = movies_df.merge(ratings_df,how = 'inner'
                                , left_on='movieId',right_on='movieId')
```

```python
movies_top_df = movies_ra_df.copy()
movies_top_df = movies_top_df[movies_top_df['rating'] >= 5.0 ]
movies_top_df = movies_top_df.groupby('title')['rating'].count().sort_values(ascending=False).head(1)
movies_top_df
```

## Задание 2

По данным файла power.csv посчитайте суммарное потребление стран Прибалтики (Латвия, Литва и Эстония) категорий 4, 12 и 21 за период с 2005 по 2010 год. Не учитывайте в расчётах отрицательные значения quantity.

```python
power_df = pd.read_csv('Файлы к заданию №2/power.csv')
transactions_df = pd.read_csv('Файлы к заданию №2/transactions.csv')
display(power_df.head(3))
display(transactions_df.head(3))
```

```python
power_work_df = power_df.copy()
#power_work_df[power_work_df['country'].str.contains('est', case=False)]['country'].unique()
power_work_df = power_work_df[
                            (power_work_df['country'].isin(['Latvia', 'Lithuania', 'Estonia']))
                            & (power_work_df['quantity'] >= 0) 
                            & (power_work_df['year'] > 2005) 
                            & (power_work_df['year'] <= 2010)
                            & (power_work_df['category'].isin([4, 12, 21]))
                            ]
power_work_df = power_work_df.groupby('country')['quantity'].sum().sort_values(ascending=False)
power_work_df
```
