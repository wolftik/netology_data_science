## Часть 2. Линейная регрессия
Обучите линейную регрессию на [датасете](https://www.kaggle.com/datasets/kumarajarshi/life-expectancy-who) ВОЗ о продолжительности жизни.

### 1. Подготовка данных
1. Скачайте датасет.
2. Импортируйте необходимые библиотеки.
3. Найдите целевую переменную (`life expectancy`).
4. Выберите ряд переменных в качестве признаков.
5. Проверьте данные на наличие пропусков и удалите их, так как модель линейной регрессии не умеет работать с пропусками.

```python
# Импортируйте необходимые библиотеки
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression # импорт модели
model = LinearRegression() # инициализация модели

# Загрузите датасет и подготовьте данные (пропуски, выбор признаков)
df = pd.read_csv("Life Expectancy Data.csv")
df.columns = df.columns.str.strip()
#df.describe()
#df.isna().sum()
df = df.dropna()
numeric_cols = df.describe().columns
# df.dtypes
# features= [col for col in numeric_cols if col not in ['Year','Life expectancy']]
features= [col for col in numeric_cols if col in ['Schooling', 'Adult Mortality', 'HIV/AIDS', 'BMI', 'Income composition of resources']]
y=df['Life expectancy']
x=df[features]
```

```python
corr = df.corr(numeric_only=True)['Life expectancy'].sort_values(ascending=False)
print('Корреляция с Life expectancy:')
for col, val in corr.items():
    print(f'  {col:40s} {val:+.4f}')
```

### 2. Обучение модели и оценка
6. Обучите модель линейной регрессии.
7. Выведите коэффициенты модели.
8. Выведите intercept (константу).
9. Для визуализации прямой выберите один наиболее значимый признак и постройте график зависимости целевой переменной от него.

```python
# Обучите модель
model.fit(x, y)

# Выведите коэффициенты и intercept
coefs = model.coef_
const = model.intercept_
print('coefs:', coefs)
print('const:', const)

for feature, coef in zip(features, coefs):
    print(f"{feature}: {coef:.4f}")
```

```python
feature = 'Income composition of resources'

# Визуализируйте зависимость для одного признака
x_range = np.linspace(df[feature].min(), 
                       df[feature].max(), 100)


x_other = df[features].drop(columns=[feature]).mean().values.reshape(1, -1)
x_other_repeated = np.repeat(x_other, 100, axis=0)
x_plot = np.column_stack([x_range, x_other_repeated])
y_plot = model.predict(x_plot)


plt.figure(figsize=(10, 6))
plt.scatter(df[feature], y, alpha=0.3, label='Данные')
plt.plot(x_range, y_plot, color='red', linewidth=2, label='Линия регрессии')
plt.xlabel(feature)
plt.ylabel('Life expectancy')
plt.title(f'Зависимость продолжительности жизни от {feature}')
plt.legend()
plt.show()
```

### 3. Преобразование признаков (полиномизация)
10. Приведите переменные к необходимой размерности и усложните их при помощи полиномизации (модуль `PolynomialFeatures`).
Выберите ряд значений для возведения в степень и визуализируйте результаты. На основе графика определите оптимальную степень полинома.

```python
# Примените PolynomialFeatures
from sklearn.preprocessing import PolynomialFeatures
X_single = df[[feature]].values  # или .values.reshape(-1, 1)

X_sorted = np.sort(X_single, axis=0)

plt.figure(figsize=(12, 4))

# Обучите модель на полиномиальных признаках и визуализируйте результат
for i, degree in enumerate([1, 3, 6, 12]):
    poly = PolynomialFeatures(degree=degree)
    X_poly = poly.fit_transform(X_single)
    
    model_poly = LinearRegression()
    model_poly.fit(X_poly, y)
    
    X_sorted_poly = poly.transform(X_sorted)
    y_pred = model_poly.predict(X_sorted_poly)
    
    # Рисуем
    plt.subplot(1, 4, i+1)
    plt.scatter(X_single, y, alpha=0.3, s=5)
    plt.plot(X_sorted, y_pred, color='red', linewidth=2)
    plt.title(f'Степень {degree}')
    plt.xlabel(feature)
    plt.ylabel('Life expectancy')

plt.tight_layout()
plt.show()
y_pred
```
